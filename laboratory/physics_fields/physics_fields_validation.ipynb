# Physics Fields Validation

This file shows validation of the physical fields visually. The pressure, the adhesion energy, the solid velocity, and mass flux fields are all displayed here.

```python
import sys

sys.path.append('/Users/rileymcnamara/CODE/2025/SCIE3212/')

from src.growkit.Simulator import TumorGrowthSimulator
from src.growkit.PlotEngine.PhysicsFieldsPlotter import PhysicsFieldsPlotter

simulator = TumorGrowthSimulator("/Users/rileymcnamara/CODE/2025/SCIE3212/templates/test_templates/physics_fields_analysis.yaml")
simulator.run()

plotter = PhysicsFieldsPlotter(simulator.field_manager)
```

### Here is the total tumour density for reference

```python
plotter.plot_tumor_density(step=0,  cmap='BuPu')
```

## Pressure Scalar Field

Plot of a slice through the centre of a circular tumour, of the pressure surface.

Solution of the pressure equation, computed with scipy conjugate gradient solver.

$$
-\nabla^2p = S_T - \nabla \cdot \left ( \frac{\delta E}{\delta\varphi_i}\nabla\varphi_T \right )
$$

```python
plotter.plot_pressure_field(step=0, cmap='plasma',)
```

## Solid Velocity Vector Field

Plot of a slice through the centre of a circular tumour, of the vector field given by the equation:

$$
\mathbf{u_s}= -\left (\nabla p - \frac{\delta E}{\delta \varphi_i}\nabla\varphi_T \right )
$$

```python
plotter.plot_velocity_field(step=0, skip=7, scale=10, width=0.005, cmap='GnBu', add_boundary_contours=True, tumor_boundary_level=0.5, boundary_color='darkblue', tumor_boundary_color='black')
plotter.plot_velocity_field(step=0, skip=7, scale=10, width=0.005, cmap='GnBu', add_boundary_contours=True, tumor_boundary_level=0.5, boundary_color='darkblue', tumor_boundary_color='black', center_x=65, center_y=65, zoom_factor=5)
```

## Adhesion Energy Scalar Field

Plot through the centre of a circular tumour, of the adhesion energy scalar field given by the functional:

$$
E = m\int_\Omega \frac{1}{4}\varphi_T^2(1-\varphi_T)^2 + |\nabla\varphi_T|^2dx
$$

```python
plotter.plot_energy_derivative_field(step=0, add_boundary_contours=True, tumor_boundary_level=0.5, boundary_color='darkblue', tumor_boundary_color='black')
plotter.plot_energy_derivative_field(step=0, add_boundary_contours=True, tumor_boundary_level=0.5, boundary_color='darkblue', tumor_boundary_color='black', center_x=65, center_y=65, zoom_factor=5)
```

## Mass Flux Vector Field

Plot through the centre of a circular tumour, of the Mass Flux vector field:

$$
J_i =-M_i\left ( \frac{\delta E}{\delta\varphi_T}\right )
$$

```python
plotter.plot_mass_flux_field(step=0, skip=1, scale=15, width=0.005, cmap='BuPu', add_boundary_contours=True, tumor_boundary_level=0.5, boundary_color='darkblue', tumor_boundary_color='black')
```
